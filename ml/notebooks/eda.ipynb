# Exploratory Data Analysis (EDA) - Bank Customer Churn
This notebook explores the Kaggle Bank Customer Churn dataset. 
Our goal is to understand the features, detect correlations, and prepare the data for training with MLflow.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-darkgrid')
```

## 1. Data Loading


```
df = pd.read_csv('../data/raw/Churn_Modelling.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

## 2. Basic Information & Missing Values


```
df.info()
print("\nMissing values:")
print(df.isnull().sum())
```

## 3. Exploratory Data Analysis
Let's look at the target variable: `Exited` (Churn)


```
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Exited', palette='Set2')
plt.title('Target Variable Distribution (Exited)')
plt.show()

churn_rate = df['Exited'].mean()
print(f"Overall Churn Rate: {churn_rate:.2%}")
```

### Feature Distributions


```
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(data=df, x='Age', hue='Exited', kde=True, ax=axes[0,0], palette='Set2')
axes[0,0].set_title('Age Distribution by Churn')

sns.histplot(data=df, x='Balance', hue='Exited', kde=True, ax=axes[0,1], palette='Set2')
axes[0,1].set_title('Balance Distribution by Churn')

sns.countplot(data=df, x='Geography', hue='Exited', ax=axes[1,0], palette='Set2')
axes[1,0].set_title('Geography Distribution by Churn')

sns.countplot(data=df, x='Gender', hue='Exited', ax=axes[1,1], palette='Set2')
axes[1,1].set_title('Gender Distribution by Churn')

plt.tight_layout()
plt.show()
```

### Correlation Matrix


```
# Drop non-numeric and unneeded columns for correlation
numeric_df = df.drop(columns=['RowNumber', 'CustomerId', 'Surname', 'Geography', 'Gender'])

plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```

## 4. Feature Engineering (Prep for ML)
We will encode categorical variables (Geography, Gender) and drop unnecessary identifiers.


```
def prep_data(df):
    df = df.copy()
    # Drop identifiers
    df = df.drop(columns=['RowNumber', 'CustomerId', 'Surname'])
    # Encode categorical
    df = pd.get_dummies(df, columns=['Geography', 'Gender'], drop_first=True)
    return df

df_prep = prep_data(df)
df_prep.to_csv('../data/processed/Churn_Modelling_Prep.csv', index=False)
print("Saved preprocessed data to data/processed/Churn_Modelling_Prep.csv")
df_prep.head()
```
